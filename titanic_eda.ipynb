## 5. Survival Analysis

```python
df.groupby('Sex')['Survived'].mean() * 100
```

```python
df.groupby('Pclass')['Survived'].mean() * 100
```

```python
df.groupby(['Sex','Pclass'])['Survived'].mean() * 100
```

```python
df.groupby(['Survived','Sex'])['Age'].median()
```

```python
df.groupby('Survived')['Age'].median()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df["FamilySize"].value_counts().sort_index()
```

```python
df.groupby('FamilySize')['Survived'].mean() * 100
```

```python
df["FamilySize"].value_counts().sort_index()
```

### Conclusions

1. Women had a substantially higher survival rate than men.

2. Survival rate was highest among first-class passengers and
   lowest among third-class passengers.

3. Female passengers in first and second class had similarly high
   survival rates. Male passengers had considerably lower survival
   rates across all classes.

4. The median age was 28 years for both survivors and non-survivors,
   so median age alone does not show a clear difference between them.

5. Most passengers traveled without family members. FamilySize = 4
   had the highest observed survival rate, although some family-size
   groups contain relatively few passengers, so these percentages
   should be interpreted carefully.
